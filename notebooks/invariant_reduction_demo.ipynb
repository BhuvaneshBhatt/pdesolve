# Invariant reduction

This notebook shows how to enumerate reduction candidates and solve through invariant reduction.

```python
import sympy as sp
import pdesolve as rle

x, t = sp.symbols('x t', real=True)
u = sp.Function('u')
eq = sp.Eq(sp.diff(u(x, t), t) + u(x, t)*sp.diff(u(x, t), x), 0)
eq
```

## Candidate enumeration

```python
candidates = rle.enumerate_invariant_reduction_candidates(eq, u(x, t), (x, t))
candidates
```

## Solve via invariant reduction

```python
result = rle.solve_via_invariant_reduction(eq, u(x, t), (x, t))
result
```
