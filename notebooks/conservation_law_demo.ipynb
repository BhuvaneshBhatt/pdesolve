# Conservation laws and Burgers equations

This notebook uses an inviscid Burgers-type PDE to illustrate recognition and explicit conservation-law routing.

```python
import sympy as sp
import pdesolve as rle

x, t = sp.symbols('x t', real=True)
u = sp.Function('u')
eq = sp.Eq(sp.diff(u(x, t), t) + u(x, t)*sp.diff(u(x, t), x), 0)
eq
```

```python
analysis = rle.analyze_first_order_nonlinear_pde(eq, u(x, t), (x, t))
analysis.burgers_family, analysis.recommended_methods
```

## Explicit conservation-law solve request

```python
record = rle.pdesolve(eq, u(x, t), (x, t), method='conservation_law')
record
```

```python
rle.summarize_solution_record(record)
```

```python
rle.extract_solution_trace(record)
```
