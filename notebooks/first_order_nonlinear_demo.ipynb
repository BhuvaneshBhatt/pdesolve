# Nonlinear first-order PDEs

This notebook illustrates nonlinear first-order analysis for a representative PDE.

```python
import sympy as sp
import pdesolve as rle

x, t = sp.symbols('x t', real=True)
u = sp.Function('u')
eq = sp.Eq(sp.diff(u(x, t), t) + u(x, t)*sp.diff(u(x, t), x), 0)
eq
```

## Structural analysis

The analysis object helps identify whether the PDE is first-order, linear or quasilinear, and whether it matches a recognized family.

```python
analysis = rle.analyze_first_order_nonlinear_pde(eq, u(x, t), (x, t))
analysis
```

## Canonical auto-dispatch

The canonical dispatcher returns a normalized solution record together with summary and trace information.

```python
record = rle.pdesolve(eq, u(x, t), (x, t), method='auto')
rle.summarize_solution_record(record)
```

```python
trace = rle.extract_solution_trace(record)
trace
```

## Direct nonlinear auto solver

```python
result = rle.solve_first_order_nonlinear_auto(eq, u(x, t), (x, t))
result
```
